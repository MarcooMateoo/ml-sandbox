## Objective
Observe how a reinforcement learning **agent** interacts with an **environment** and receives **rewards** based on its actions.

> **Note:** This notebook does **not** implement a reinforcement learning algorithm. Instead, it demonstrates the reinforcement learning interaction cycle discussed in the module.


## Scenario

A recommendation system suggests three articles to a user.

For each recommendation:
- The **agent** chooses an action (which article to recommend).
- The **environment** (the user) provides feedback.
- The agent records the reward.

In real reinforcement learning systems, this feedback is used to improve future decisions.


````python
# Sample interaction between an agent and its environment

actions = [
    "Recommend Article A",
    "Recommend Article B",
    "Recommend Article C"
]

rewards = [5, 0, -1]

print("Agent Interaction\n")

total_reward = 0

for action, reward in zip(actions, rewards):
    print(f"Action: {action}")
    print(f"Reward received: {reward}\n")
    total_reward += reward

print(f"Total Reward = {total_reward}")
````
